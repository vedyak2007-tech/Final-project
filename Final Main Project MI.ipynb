```python
data = {
    "Name":[ "Priya","Harini","Divya","Sneha","Kavya","Nithya","Swetha","Keerthana","Ananya","Ramya","Sanjana","Madhu"],
    "Study_Hours": [2, 3, 5, 1, 6, 4, 2, 7, 3, 8, 5, 1],
    "Attendance": [60, 65, 75, 40, 85, 70, 50, 90, 55, 95, 80, 35],
    "Previous_Marks": [45, 50, 60, 30, 70, 55, 35, 80, 40, 85, 65, 25],
    "Final_Marks": [48, 52, 65, 35, 78, 60, 40, 88, 45, 92, 72, 30],
    "Result": ["Fail", "Pass", "Pass", "Fail", "Pass", "Pass",
               "Fail", "Pass", "Fail", "Pass", "Pass", "Fail"]
}
df = pd.DataFrame(data)
print(df)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
X = df[["Study_Hours", "Attendance", "Previous_Marks"]]
y = df["Final_Marks"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)
model = LinearRegression()
model.fit(X_train, y_train)
print("Linear Regression Model Trained Successfully")
print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)
print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
```

```python
student_name = input("Enter Student Name: ")
student = df[df["Name"].str.lower() == student_name.lower()].iloc[0]
predicted_marks = model.predict(
    student[["Study_Hours", "Attendance", "Previous_Marks"]].to_frame().T
)
print("\nStudent Name:", student["Name"])
print("Study Hours:", student["Study_Hours"])
print("Attendance:", student["Attendance"])
print("Previous Marks:", student["Previous_Marks"])
print("Predicted Final Marks:", round(predicted_marks[0], 2))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
print("Classification libraries imported successfully")
```

```python

X = df[["Study_Hours", "Attendance", "Previous_Marks"]]
y = df["Result"]
print("Input Features:")
print(X.head())
print("\nTarget Variable:")
print(y.head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)
print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
```

```python
classifier = LogisticRegression(max_iter=1000)
classifier.fit(X_train, y_train)
print("Logistic Regression Model Trained Successfully")
```

```python
new_student = pd.DataFrame({
    "Study_Hours": [5],
    "Attendance": [80],
    "Previous_Marks": [65]
})
predicted_result = classifier.predict(new_student)
print("Predicted Student Result:", predicted_result[0])
```

```python

name = input("Enter Student Name: ")
hours = float(input("Enter Study Hours: "))
attendance = float(input("Enter Attendance (%): "))
previous = float(input("Enter Previous Marks: "))
new_student = pd.DataFrame({
    "Study_Hours": [hours],
    "Attendance": [attendance],
    "Previous_Marks": [previous]
})
marks = model.predict(new_student)[0]
result = classifier.predict(new_student)[0]
print("\n===== STUDENT PERFORMANCE REPORT =====")
print("Student Name:", name)
print("Study Hours:", hours)
print("Attendance:", attendance)
print("Previous Marks:", previous)
print("Predicted Final Marks:", round(marks, 2))
print("Predicted Result:", result)
```
